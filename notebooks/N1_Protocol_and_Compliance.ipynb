Notebook ini memverifikasi kepatuhan sistem terhadap 16 aturan permanen (R1–R16), enam kondisi eksperimen (K0, K1, K2, K3, K6, KP), serta hashing manifest SHA-256.

```
import hashlib
import json
from pathlib import Path

# Verifikasi checksum manifest & ledger immutability
def verify_file_sha256(file_path: Path) -> str:
    hasher = hashlib.sha256()
    with open(file_path, 'rb') as f:
        hasher.update(f.read())
    return hasher.hexdigest()

print('Protokol Kepatuhan SHADOW Terkunci:')
print('- R1: Kepemilikan PRD sebagai sumber kebenaran tunggal')
print('- R2: Pemisahan fisik brankas uji tersegel (Sealed Holdout)')
print('- R4: Fit-di-dalam-lipatan ketat (No leakage)')
print('- R5: Pencatatan append-only ke ledger.jsonl')
print('- R12: 100% eksekusi lokal tanpa dependensi API eksternal')
```
