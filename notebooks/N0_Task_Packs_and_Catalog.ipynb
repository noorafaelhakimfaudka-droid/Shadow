### Taksonomi 24 Kesalahan Tertanam (Defect Catalog)

| ID | Kategori | Nama Cacat | Deskripsi Operasional | Dampak Evaluasi |
| :--- | :--- | :--- | :--- | :--- |
| **D01 (L1)** | Kebocoran Target | Salinan Langsung Target | Fitur disalin dari target + derau mikro | Akurasi artifisial 99% |
| **D02 (L2)** | Kebocoran Target | Proksi Deterministik | Kolom pasca-event penentu label | Validasi bocor |
| **D03 (L3)** | Kebocoran Agregasi | Agregasi Global Sebelum Split | Mean target dihitung sebelum fold | Optimistic bias |
| **D04 (L4)** | Kebocoran Prosedur | Penskalaan Global | StandardScaler di-fit pada seluruh data | Contamination |
| **D05 (L5)** | Kebocoran Waktu | Pembalikan Temporal | Data masa depan dipakai prediksi masa lalu | Lookahead bias |
| **D06 (L6)** | Kebocoran Grup | Pemecahan Entitas Cluster | Entitas grup yang sama ada di train & test | Overfitting tersembunyi |
| **D07 (L7)** | Kontaminasi | Duplikasi Baris Lintas Split | Sampel identik di train dan test | Skor menggelembung |
| **D08 (L8)** | Kebocoran Indeks | Korelasi ID Baris | ID terurut berkorelasi kuat dengan kelas | Indeks bocor |
| **D09 (L9)** | Kebocoran Status | Flag Penghentian Proses | Kolom status proses hanya ada saat positif | Label leakage |
| **D10 (L10)**| Kebocoran Seleksi | Feature Selection Pra-CV | Memilih 10 fitur terbaik sebelum K-Fold | Selection bias |
| **D11–D24** | Evaluasi & Metrik | Cacat Metrik & Imbalance | ROC-AUC pada data 99:1 tanpa PR-AUC, dsb | Kesimpulan keliru |
