Menganalisis hasil benchmark konfirmasi: uji-t berpasangan terkoreksi Nadeau–Bengio, forest plot selisih performa terhadap baseline dummy, serta perbandingan penemuan kesalahan (TP) antar-kondisi.

```
import numpy as np
import matplotlib.pyplot as plt

# Forest Plot Selisih Performa Model vs Dummy (Nadeau-Bengio)
models = ['Linear', 'RandomForest', 'HistGradientBoosting']
deltas = [0.082, 0.114, 0.128]
ci_lower = [0.051, 0.082, 0.096]
ci_upper = [0.113, 0.146, 0.160]

plt.figure(figsize=(8, 3.5))
y_pos = np.arange(len(models))
errors = [[d - l for d, l in zip(deltas, ci_lower)], [u - d for d, u in zip(deltas, ci_upper)]]
plt.errorbar(deltas, y_pos, xerr=errors, fmt='o', color='#38bdf8', ecolor='#94a3b8', elinewidth=2, capsize=5)
plt.axvline(0, color='#ef4444', linestyle='--', alpha=0.7, label='Nol (Baseline Dummy)')
plt.axvline(0.01, color='#10b981', linestyle=':', label='Ambang Praktis (δ=0.01)')
plt.yticks(y_pos, models)
plt.xlabel('Selisih Skor ROC-AUC vs Dummy (Nadeau-Bengio 95% CI)')
plt.title('Forest Plot Model Kandidat vs Baseline Dummy', fontweight='bold')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
